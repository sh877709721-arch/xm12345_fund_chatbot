```python
import re
# 清洗文本 (简单去除非中文字符，假设无分词库，用空格分隔)
def clean_text(text):
    text = re.sub(r'[^\u4e00-\u9fa5]', ' ', text)  # 保留中文，用空格替换其他
    text = ' '.join(text.split())  # 去除多余空格
    return text

df = final_document

df['clean_text'] = df['text'].apply(clean_text)

# 由于环境限制，无法加载bge-m3或LLM，使用简单词频初筛 (计算词数作为信息量代理，阈值>5词)
df['word_count'] = df['clean_text'].apply(lambda x: len(x.split()))
df_filtered = df[df['word_count'] > 5].copy()

# 话题归类：使用现有'诉求小类'作为话题，或简单groupby
topic_summary = df_filtered.groupby('诉求小类').size().reset_index(name='count')
topic_summary = topic_summary.sort_values('count', ascending=False)

# 提取热点：Top话题及其代表描述
hotspots = []
for idx, row in topic_summary.iterrows():
    topic = row['诉求小类']
    count = row['count']
    sample_texts = df_filtered[df_filtered['诉求小类'] == topic]['text'].head(2).tolist()  # 前2个示例
    hotspots.append({'topic': topic, 'count': count, 'samples': sample_texts})

print("Filtered DataFrame:")
print(df_filtered[['id', '诉求小类', 'clean_text']])
print("\nTopic Summary:")
print(topic_summary)
print("\nHotspots:")
print(hotspots)
```
